```python
#csv file handling 
# creating a csv file
#used csv lib 
import csv 
with open('krish.csv','w', newline='') as f: #new line parameter will remove the blanks from this
    w=csv.writer(f)
    w.writerow(['name','age','salary'])
    w.writerow(['karan',21,0])
    w.writerow(['Niraj',22,100000])
    w.writerow(['Manish',23,50000])
```

```python
#reading a csv file
import csv
with open('krish.csv','r') as f:
    r=csv.reader(f)
    data=list(r) #converting to list
    print(r)
    print(data)
```

```python
#to display each element of csv individualy 
for i in data:
    for j in i:
        print(j, end=' ') #space btw elements
    print()               #new row on new line
```
