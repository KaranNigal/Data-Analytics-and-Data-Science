```python
# try:
#     risky code

# except:
#     handling code

# finally:
#     cleanup code
```

```python
print('welcome')
try:
    n1=int(input('enter number 1'))
    n2=int(input('enter number 2'))
    print(n1/n2)
except ZeroDivisionError as msg:
    print('exception: ',msg)
finally:
    print('task completed')
```

```python
#else block is run when there is no exception found
print('welcome')
try:
    n1=int(input('enter number 1'))
    n2=int(input('enter number 2'))
    print(n1/n2)
except ZeroDivisionError as msg:
    print('exception: ',msg)
else:
    print('i am else block no exception in code')
finally:
    print('task completed')
```

```python
# Various possible combination of try_except_else_finally:
# 1. try -- error
# 2. except --error
# 3. else --error
# 4. finally --error
# 5. try-exception --no error
# 6. try-else --erro
# 7. try-finally --no error
# 8. try_except_else_finally --no error >> Recommended
# 9. try-else-except-finally --error
```

```python
# we need to keep closing file
# we need not to close file if we use with open
with open('Maverick.text','r') as f:
    print(f.read())
    print('inside with open',f.closed)   #file will still be open
print('outside with open',f.closed)      #file automatically closed outside with open
```

```python
# read and write mode 
# also to check the current read cursor position
# also to add something by moving cursor anywhere
with open('Maverick.text','r+') as f:
    print(f.read())
    print(f.tell()) #tells the current position of read cursor
    f.seek(0)       #moves the cursor to desired position
    f.write("hi")   # it will replace ka with hi
    print(f.tell())
    f.seek(0)
    print(f.read()) #it will read from current cursor postion thats why we seek cursor to 0 in prev step
```

```python
#write and read mode
with  open('Maverick.text','w+') as f:
    print(f.tell())
    f.write('Karan is the best')
    print(f.tell())
    f.seek(0)
    print(f.read())
```

```python
#append and read mode
with  open('Maverick.text','a+') as f:
    print(f.tell())
    f.write('   AJ is the best') # this will get appendend not overiden bcz of a+ mode 
    print(f.tell())
    f.seek(0)
    print(f.read())
```

```python
# now implementing exception handling with file handling , specifically for finally block demo
print('now opening file')
try:
    f=open('Maverick.text','r')             #risky code in try
    print(f.read())
except FileNotFoundError as msg:
    print('exception',msg)                  # handling code in except
finally:
    f.close()                              # closing code in finally block
    print('file closed')
```
