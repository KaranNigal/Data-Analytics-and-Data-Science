```python
# duplicate keys not aloowed but value can be duplicate
#can store multiple data  type
# insertion order preserved 
# no indexing and slicing 
```

```python
# can store various data types
a={}
a["eid"]= 100
a["ename"]= "karan"
a["esal"]=0
a[100]= "20"
print(a)
```

```python
# accesing a value, with help of key
a={'eid': 100, 'ename': 'karan', 'esal': 0, 100: '20'}
a['ename']
```

```python
# updating a dict
a['dept']='it'
print(a)
```

```python
#delete a element
b={'eid': 100, 'ename': 'karan', 'esal': 0, 100: '20', 'dept': 'it'}
del b['dept']
print(b)
```

```python
# to delete a entire dict
b.clear()
print(b)
```

```python
# delete enitre dictionary
b={}
del b
# print(b) return error not var 
```

```python
# get method to get value associated ith a key
b={'eid': 100, 'ename': 'karan', 'esal': 0, 100: '20', 'dept': 'it'}
b.get('dept')
```

```python
# example of defult value in get method
c={'eid': 100, 'ename': 'karan', 'esal': 0, 100: '20'}
c.get('dept','mech') #this key dept's value is not presernt so default value mech returned
```

```python
#but now lets try for key that is present and has value
c.get('ename','aditi')
```

```python
#pop method 
c.pop('ename')
```

```python
print(c)
```

```python
#popitem method removes last element  
c.popitem()
print(c)
```

```python
c={'eid': 100, 'ename': 'karan', 'esal': 0, 100: '20'}
# keys methods return all the  keys
c.keys()
```

```python
# values method returns all the values
c.values()
```

```python
# items method to get all content
c.items()
```

```python
# set default method
c.setdefault("dept","mech") #if no key value then added in dict
print(c)
```

```python
c.setdefault('eid',105)
# checks if key present, if present leave as it is
```

```python
# update method in dict
# add one dict in other
c={'eid': 100, 'ename': 'karan', 'esal': 0, 100: '20'}
d={1:101, 2:202, 3:303}
c.update(d)
print(c)
```

```python
# fromkeys method
b=dict.fromkeys("maveirck",2)
print(b)
```

```python
# String format
a=10
b=20
c= a+b
print("the addition of {} and {} is {}".format(a,b,c))
```
