```python
# There are three types of statement in pyhton
# 1. Conditional statement eg. if , if else , if elif
# 2. Iterative st eg. for , while loop
# 3. Transfer st eg. pass, continue, break
```

```python
# ============================================================
# CONDITIONAL STATEMENTS
# ============================================================

# Used to execute code based on a condition.

# ------------------------------------------------------------
# 1. if
# ------------------------------------------------------------

# Syntax:
# if condition:
#     statements

# Runs only if the condition is True.


# ------------------------------------------------------------
# 2. if - else
# ------------------------------------------------------------

# Syntax:
# if condition:
#     statements
# else:
#     statements

# if  -> runs when condition is True
# else -> runs when condition is False


# ------------------------------------------------------------
# 3. if - elif - else
# ------------------------------------------------------------

# Syntax:
# if condition1:
#     statements
# elif condition2:
#     statements
# else:
#     statements

# elif = "else if"
# Used to check multiple conditions.
# First True condition is executed.


# ------------------------------------------------------------
# 4. Nested if
# ------------------------------------------------------------

# Syntax:
# if condition1:
#     if condition2:
#         statements

# An if statement inside another if statement.


# ------------------------------------------------------------
# 5. Comparison Operators
# ------------------------------------------------------------

# ==   Equal to
# !=   Not equal to
# >    Greater than
# <    Less than
# >=   Greater than or equal to
# <=   Less than or equal to


# ------------------------------------------------------------
# 6. Logical Operators
# ------------------------------------------------------------

# and -> Both conditions must be True
# or  -> At least one condition must be True
# not -> Reverses True/False


# ------------------------------------------------------------
# 7. Ternary / Short-hand if-else
# ------------------------------------------------------------

# Syntax:
# value_if_true if condition else value_if_false


# ------------------------------------------------------------
# IMPORTANT
# ------------------------------------------------------------

# ':' is compulsory after if, elif and else.
# Indentation is compulsory.
# '='  -> assignment
# '==' -> comparison
# else has no condition.
# Multiple elif statements are allowed.
# Only one else is allowed in an if-elif chain.
```

```python
# check if a number is postive , negative or zero 
b=int(input('enter the no to check'))
if b>0:
    print('no is positive')
if b==0:
    print('no is zero')
if b<0:
    print('no is negative')
```

```python
# check if the number is even or odd 
a= int(input('enter the no to check even odd'))
if a%2==0:
    print('even')
else:
    print('odd')
```

```python
# check a person is eligible to vote 
age= int(input('enter your age'))
if(age>=18):
    print('you are eligible to vote')
else:
    print('you are not eligible to vote')
```

```python
# check whater given char is vovel or consonant
vowel=['a','e','i','o','u']
c= input('enter the aplhabet').lower()
if c in vovel:
    print('char is vowel')
else:
    print('char is consonant')
```

```python
# wrp to check wheater given string is palindrome or not 
s= input('enter the string to check palindrome')
if(s==s[::-1]):
    print('palindrome')
else:
    print('not palindrome')
```

```python
# check is a number is divisible by 5 and 7 and give output accordingly
num = int(input('Enter the number'))
if num%5==0 and num%7==0:
    print(f'{num}number is divisible by both 5 and 7')
elif num%5==0 and num%7!=0:
    print(f'{num}number is divisible by 5 only')
elif num%5!=0 and num%7==0:
    print(f'{num}number is divisible by 7 only')
else:
    print(f'{num}number is not divisible by both 5 and 7')
```

```python
# make a grade calculator 
percent= int(input('enter your percentage'))
if percent>=0 and percent<=100: # concept of nested if else used for input validation 
    if percent>80:
        print('A grade')
    elif percent >65:
        print('B grade')
    elif percent >35:
        print('C grade')
    else:
        print('Congrats !! you failed')
else:
    print('enter valid percent')
```
