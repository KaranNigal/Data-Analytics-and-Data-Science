```python
# filter function 
# used to apply the filter to a sequnce and create new sequence 
# syntax-- filter(fn,seq)
# to filter even numbers from a range 
list(filter(lambda x: x%2==0, range(1,20)))
```

```python
# filter out strings with length greater than 3
words=['karan','aditi','aya','hi']
list(filter(lambda x: len(x)>3, words))
```

```python
# filter positive numbers from a list 
numbers=[1,2,3,-2,-3,-4]
list(filter(lambda x: x>0 , numbers))
```

```python
# filtering odd numbers from a list 
list(filter(lambda x: x%2!=0, range(1,21)))
```

```python
# filtering numbers divisible by 5
list(filter(lambda x: x%5==0, range(1,21)))
```

```python
# Reduce function 
from functools import reduce 
numbers= [1,2,3,4,5]
p= reduce(lambda x,y: x+y, numbers)
print(p)
```

```python
# to find factorial of a number 
n= int(input('enter the number'))
fact= reduce(lambda x,y: x*y, range(1,n+1))
print(fact)
```

```python
# product of all elements in a list  
a=[1,2,3,4,5]
sum= reduce(lambda x,y: x*y , a )
print(sum)
```

```python
# to find the largest element in a list
a=[11,24,67,89,101]
largest= reduce(lambda x,y: x if x>y else y, a)
print(largest)
```

```python
# Flatening of a list 
# to convert nested lists into 1d list using reduce function
lol=[[1,2],[3,4],[5,6]]
li= reduce (lambda x,y: x+y, lol) # concatenation of lists
print(li)
```

```python
class Student :                            #class creation
    def __init__(self,name,marks):         #constructor
        self.name= name                    #instance variables
        self.marks= marks

    def display(self):                     #instance method
        print('name of student : ',self.name) 
        print('marks of student : ',self.marks)

s=Student('karan',31)                     #object creation, here s is reference variable that refers the created object
s.display()                               # object calling
```

```python
# insatnce variable can be delcared inside constructor, instance methods, and outside class using object reference
class test:
    def __init__(self):
        self.a=10     # var inside constructor
    def m1(self):      
        self.b=20     # var inside method

t= test()           
t.m1()               # run instance to declare the b
t.c=30               # outside class using reference of t object
print(t.__dict__)    # __dict__ function display the information of created object
```

```python
# we can acess the instance variable using the instance  methods or outside class using obejct 
class example:
    def __init__(self):
        self.a= 10
        self.b= 20
        self.c= 30

    def m(self):                    #variable acess using method 
        print(self.a)

t= example()
t.m()                               #outside class using obejct reference 
print(t.c)
```
