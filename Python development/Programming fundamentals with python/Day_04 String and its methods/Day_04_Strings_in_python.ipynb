```python
#startswith() and endswith() method in strings
e="karannigal69@gmail.com"
print(e.startswith("karan"))
print(e.endswith("nigal"))
print(e.endswith("com"))
```

```python
#methods in string to change the case of string
k="mavErick is tHe beSt" 
print(k.upper()) #all into upper case
print(k.lower()) #all into lower case
print(k.swapcase()) #all case get swapped, lower to upper , upper to lower
print(k.title()) # only first character of the string gets capitalized
print(k.capitalize()) #only the first character gets capitalized
```
