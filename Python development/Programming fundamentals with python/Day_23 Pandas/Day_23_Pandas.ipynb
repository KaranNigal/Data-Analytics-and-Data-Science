```python
# Concatenation of data frames
```

```python
dict1={'id':[464,868,105,979,2424],
       'name':['aab','fadd','dad','dfasd','dffg'],
       'salary':[35000,40000,15000,10000,16000]}
df1=pd.DataFrame(dict1)
df1
```

```python
dict2={'id':[101,102,103,104,105],
       'name':['jon','setu','vijay','allu','teja'],
       'salary':[12000,22000,28000,18000,16000]}
df2=pd.DataFrame(dict2)
df2
```

```python
df3=pd.concat([df1,df2],ignore_index=True)

df3
```

```python
# Merge in pandas similar to join in sql
# In pandas, the merge function is used to combine two DataFrames based on a key (or multiple keys).

# It is similar to SQL joins and allows you to perform various types of joins (inner, left, right, outer) to combine rows from different DataFrames.
```

```python
#df1
dict1={'id':[101,301,300],
       'name':['babita','sangita','ranjita'],
       'salary':[35000,40000,15000]}
df1=pd.DataFrame(dict1)
df1
```

```python
#df2
dict2={'id':[101,102,300,205],
       'name':['akash','suresh','mohit','nilesh'],
       'salary':[12000,22000,28000,18000]}
df2=pd.DataFrame(dict2)
df2
```

```python
# inner merge: showing only common ids
pd.merge(df1,df2,on='id',how='inner')
```

```python
# left merge :
# all records from left dataframe commen records from right
```

```python
df3=pd.merge(df1,df2,on='id',how='left')
df3
```

```python
# right merge:
# comman records from left all records from right
```

```python
df3=pd.merge(df1,df2,on='id',how='right')
df3
```

```python
#experiment : nested inner join
pd.merge(df3,pd.merge(df1,df2,on='id',how='inner'),on='id',how='inner')
```

```python
df3=pd.merge(df1,df2,on='id',how='outer')
df3
```

```python
# Handling Missing Values
```

```python
dict1={
    'Duration': [60, 60, 45, 450, 30],
    'Date': ['2020/12/01', '2020/12/01', '2020/12/03', None, '2020/12/05'],
    'Pulse': [110, 117, 109, 104, 109],
    'Calories': [409.1, None, 282.4, 253.3, None]
}

df3 = pd.DataFrame(dict1)
df3
```

```python
df3.isnull().sum() # to check null in each col
```

```python
df3.dropna() #delete the row with null values
```

```python
df3
```

```python
#replacing null values with mean in calories col
df3["Calories"]=df3["Calories"].fillna(df3["Calories"].mean())
df3
```

```python
#replacing null values with median in calories col
df3["Calories"]=df3["Calories"].fillna(df3["Calories"].median())
df3
# when there is no outliers in the data, median == mode
```

```python
# replacing null date with most repeated date ie mode
df3["Date"]=df3['Date'].fillna(df3["Date"].mode()[0]) #here 0 is mentioned bcz in tuple of modes, we select the first one (most frequent)
df3
```
