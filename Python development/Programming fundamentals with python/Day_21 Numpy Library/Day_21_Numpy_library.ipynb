```python
# What is NumPy?
# NumPy (Numerical Python) is a Python library used for numerical and scientific computing.

# Why NumPy?
# Fast and memory-efficient
# Written in C, so it is much faster than Python Lists
# Used to create N-dimensional arrays
# Supports matrix operations, linear algebra and statistical calculations
# Widely used in Data Analysis, Machine Learning and Deep Learning
# Note: ML models usually expect data in a 2D format (rows × columns).
```

```python
# traditional way using lists 
# product of two list
a=[2,4,8]
b=[4,2,3]
c=[]
for i in range(len(a)):
  c.append(a[i]*b[i])
print(c)
```

```python
# product of 2 arrays 
a=np.array([2,4,8])
b=np.array([4,2,3])
a*b
# hence the loop avoided , called vectorization
```

```python
# Vectorization
# Vectorization means performing operations on the entire array without writing explicit loops.
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
print(a+b)
print(a-b)
print(a*b)
print(a/b)
```
