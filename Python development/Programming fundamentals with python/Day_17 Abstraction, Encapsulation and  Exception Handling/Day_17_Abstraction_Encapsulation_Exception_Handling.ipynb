```python
# creating a abstract class with abstract method and a child class properly implementing both
from abc import ABC, abstractmethod

class test(ABC):      #abstract class created by inheriting ABC
    @abstractmethod   #abtract method created but it should be passed , we cant implement it here
    def m1(self):
         pass
    def color(self):  # a concrete method color also created 
        print('white')

class child(test):   # child class implementing the abstract class created 
    def m1(self):
         print('implemented the abstract method')
    def color(self): # overiden the concrete method
        print('pink')

c=child()            # the object will be created bcz we implemented abstract class and method m1 
c.m1()
c.color()
# hence the abtract class and mehtod needs to be implemented first using child class then only we can run the prog
```

```python
# creating a abstract class for car 
#then creating car model clases that implements the abstract class car

class car(ABC):
    @abstractmethod
    def mil(self):
        pass

class nexon(car):
    def mil(self):
        print(25)

class merc(car):
    def mil(self):
        print(7)

class bmw(car):
    def mil(self):
        print(10)

n= nexon()
m= merc()
b= bmw()
n.mil()
m.mil()
b.mil()
```

```python
# cretae a abstract class vehicle that comtains abstract method wheels
# make class rishaw, car, bike that print there no of wheels

from abc import *
class vehicle(ABC):
    @abstractmethod
    def wheels():          # we can use without self keyword also here 
        pass

class rikshaw(vehicle):
    def wheels(self):
        print('rikshaw has 3 wheels')

class car(vehicle):
    def wheels(self):
        print('car has 4 wheels')

class bus(vehicle):
    def wheels(self):
        print('bus has 6 wheels')

r= rikshaw()
r.wheels()
c= car()
c.wheels()
b= bus()
b.wheels()
```

```python
# declare a abstract class shape with two abstract methods: area and perimeter 
# make classes like circle, square, etc that would implement this and return area and perimeter 

class shape(ABC):
    @abstractmethod
    def area():
        pass
    @abstractmethod
    def perimeter():
        pass

class circle(shape):
    def __init__(self, radius):
        self.radius= radius
    def area(self):
        radius= self.radius
        print("area of circle is", 3.14*radius*radius)
    def perimeter(self):
        radius= self.radius
        print('perimeter of circle is', 2*3.14*radius)

class rect(shape):
    def __init__(self, l, b):
        self.l= l
        self.b= b
    def area(self):
        l=self.l
        b=self.b
        print("area of square is", l*b)
    def perimeter(self):
        l=self.l
        b=self.b
        print('perimeter of square is', 2*(l+b))

c=circle(2)
c.area()
c.perimeter()
r=rect(2,3)
r.area()
r.perimeter()
```

```python
# Access specifiers 
# public : acess outside , inside or anywhere of class (var)
# protected : accesed mostly inside the class or child class but not recommended to access outside the class ie by its objects (_var)
# private : can be accesses only inside the class and its method, the objects cant acess it (__var)
#         : Name Manglinng: python changes the name of private var internally so that it cant be access outside
#         : we can access private var outside class using name mangling , the var name changes to obj._Classname__privvar
#         : ensures security and privacy
```

```python
# demo of public var 
class test:
    def __init__(self,a):
        self.a=a             #public var declared normally
t=test(10)
t.a                         # can be accessed outside using objects
```

```python
# demo of protected var
class student:
    def __init__(self):
        self._marks= 85      #protected var delcared using _ ie single underscore before its name
    def display(self):
        print(self._marks)   #can be accessed by methods inside the class

s=student()
s._marks                     # can also be accessed outside the class using object but not recommmended 
```

```python
# demo of private var 

class student:
    def __init__(self):
        self.__marks= 85      #private var delcared using __ ie double underscore before its name
    def display(self):
        print(self.__marks)    #can be accessed by methods inside the class

s=student()
# s.__marks                 #cannot access outside the class hence returns error
s._student__marks           # but we can access the private var outside the class using name mangling
```

```python
# use of private var in banking system

class account:
    def __init__(self):
        self.__bal=100               # account bal kept private
    def deposit(self,d):
        self.__bal+=d                #can be accesed by methods inside the class 
    def show(self):
        print(self.__bal)            # method created to show the private var bal

acc=account()
acc.deposit(1000)
acc.show()
#acc.__bal                         #this will return error as bal is private and we cant acess it
```

```python
# Accessing private var revenue outside the class fin by other class hr using get method
# private var can be accessed either by declaring method in class or by name mangling

class fin:
    def __init__(self):
        self.__revenue= 100      #private var declared
    def get(self):
        print(self.__revenue)    #method to access the private var

class hr:
    def __init__(self, fin_obj):
        # print(fin_obj.__revenue) this will return error as private var cant accessed directly
        print(fin_obj.get())    #called get to display private var
f=fin()
h=hr(f)
```

```python
# Exceptional Handling
# used to handle run time errors 
print('hello')
print(10/0)      # this will create runtime error ie exeption 
print('AJ')
print('I hope you remember me')
#remaining code didnt ran successfuly
```

```python
print('hello')
try:                                      #handled the exeption using try execpt ie. exeption is handled
    print(10/0)                           #detects the error, so risky code in try block
except:
    print('exeption occured')             #handles the error, so handling code in except block
print('AJ')
print('I hope you remember me')
# remaining code ran succesfully
```

```python
# to handle specific type of exception
print('hello')
try:
    print(10/0)
except ZeroDivisionError:     #Mentioned specific type of exception 
    print('exeption occured:')   
print('AJ')
print('I hope you remember me')
```

```python
# to print the message of exeption
print('hello')
try:
    print(10/0)
except ZeroDivisionError as msg:     #we can mention as well which error to except, also used msg to print it
    print('exeption occured:',msg)   #msg will show detailed msg of exception
print('AJ')
print('I hope you remember me')
```

```python
# to handle multiple type of exception
print('name')
try:
    n1=int(input('enter no'))
    n2=int(input('enter no'))
    print(n1/n2)
except (ZeroDivisionError, ValueError) as msg:     # erors given as tuple, this will handle multiple types of exception and return message
    print("exception occured: ",msg)   
print('task completed')
```
