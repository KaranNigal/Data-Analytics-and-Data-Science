```python
#nested loop to display the elements of  nested list
a=[[1,2,3],[4,5],[6]]
for i in a:
    for j in i:
        print(j)
```

```python
#nested loop to display the elements of  nested tupple
a=((1,2,3),(4,5),(6,))
for i in a:
    for j in i:
        print(j)
```

```python
#nested loop to display the elements of  nested tuple+list
a=[(1,2,3),(4,5),(6,)]
for i in a:
    for j in i:
        print(j)
```

```python
# for loop to display the key of a dictionary
d={"a":1,'b':2,'c':3,'d':4}
for i in d:
    print(i)
# it will only display the keys 
```

```python
# # for loop to display the values of a dictionary
d={"a":1,'b':2,'c':3,'d':4}
for i in d.values():
    print(i)
```

```python
#  for loop to display the items of a dictionary
d={"a":1,'b':2,'c':3,'d':4}
for i,j in d.items():
    print(i,j)
```

```python
# print rigth angle triangle with stars
n=int(input("Enter number"))
for i in range(1,n+1):
    print("*"*i)
```

```python
# print inverted rigth angle triangle with stars
n=int(input("Enter number"))
for i in range(n,0,-1):
    print("*"*i)
```

```python
# print a proper triangle with stars for given n height
n= int(input('enter height of triangle'))
for i in range(1,n+1):
    print((' '*(n-i)) + ("* "*i))
```

```python
# print a proper inverted triangle with stars for given n height
n= int(input('enter height of triangle'))
for i in range(n,0,-1):
    print((' '*(n-i)) + ("* "*i))
```

```python
# print a proper inverted triangle with stars for given n height
n= int(input('enter height of triangle'))
for i in range(0,n):
    print((' '*i) + ("* "*(n-i)))
```

```python
# display the diamond pattern using above triangle code
n= int(input('enter height of triangle'))
for i in range(1,n):
    print((' '*(n-i)) + ("* "*i))
for i in range(0,n):
    print((' '*i) + ("* "*(n-i)))
```

```python
# Comprehension
# used to shorten the code using comphrensive statements 
# cant be used with tuple generates generator class 
# syntax -- [expression for element in some sequence] 
```

```python
# example of creating a list using comphrensive statement 
p=[x*x for x in range(1,11)]
print(p)
```

```python
# same with set
p={x**x for x in range(1,11)}
print(p)
```

```python
# creating a list of cubes for a range
p=[x**3 for x in range(1,11)]
print(p)
```

```python
# same with dictionary
p={x:x**4 for x in range(1,11)}
print(p)
```

```python
a=[1,2,3,4,5]
p=[x+5 for x in a]
print(p)
```
