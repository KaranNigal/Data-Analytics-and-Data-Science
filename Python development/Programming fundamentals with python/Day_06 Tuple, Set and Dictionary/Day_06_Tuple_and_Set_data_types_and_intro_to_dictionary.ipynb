```python
## SET data type
# each value is unique
# duplicate not allowed
# no insertion order preserved
# no indexing and slicing
# hetereogenous element allowed
# {}
```

```python
# by default dictionary
a={}
type(a)
```

```python
# type casted to set
a=set()
print(type(a))
```

```python
# set is mutable so adding new element does not changes it memory location
# add method demo
print(id(a))
a.add(45)
a.add(69)
a.add(1)
a.add(0)
a.add(99)
print(a)
print(id(a))
```

```python
# duplicate not allowed , unique stored 
# heteregenous data types
a={10,10,5,5,6,6,78,"karan"}
print(a)
```

```python
len(a)
```

```python
# update method for set
# inserts different data types in the set
a= {100,200,300}
d=[1,2,3]
e= "maverick"
f=(5,7,8)
a.update(d,e,f)
print(a)
```

```python
# remove method
a={1,2,3,4}
a.remove(4)
print(a)
```

```python
# pop method
a={1,2,3,4}
a.pop()
print(a)
```

```python
# copy method
b=a.copy()
print(b)
```

```python
# dicard method 
# removes element but if not present, does not return error
a.discard(100)
```

```python
# SET OPERATIONS
# intersection
a={1,2,3,4}
b={3,4,5}
print(a.intersection(b))
print(a)
print(b)
```

```python
# difference 
a={1,2,3,4,5,6}
b={3,4,5,6,7}
a.difference(b)
```

```python
# union
a.union(b)
```

```python
# symmetric difference 
a.symmetric_difference(b)
```

```python
# diferene update 
a.difference_update(b)
print(a)
```
