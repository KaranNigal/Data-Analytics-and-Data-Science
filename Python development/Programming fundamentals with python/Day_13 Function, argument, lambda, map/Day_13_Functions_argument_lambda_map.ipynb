```python
# function to give largest  number in a collection
def highest(n):
    max=n[0]
    for i in n:
        if i>max:
            max=i
    print(max)
highest([1,2,3,4,5])
```

```python
# function to check give number prime or not 
import math
def prime_check(n):
    if n>1:
        for i in range(2,int(n**0.5)):
            if n%i==0:
                print('number not prime')
                break
        else:
            print('prime')
    else:
        print('not prime')
prime_check(11)
```

```python
#positional argument 
def wish(name, msg):
    print("hello", name, 'very', msg)
wish('K','mrng') #here the correct output is depends on the position of parameters
wish('mrg','k') # here the function is not as expected
```

```python
# to solve the above  problems we can give the arguments with a keywords 
# Key word argument 
wish(msg='mrg',name='k') #this will solve the issue of position
```

```python
#default argument 
# incase no argument then default one is used 
# used to avoid errors and to later  implementation
def wish(name='moron'):
    print('hello',name,'how are u?')
wish('maverick')
wish()
```

```python
#variable length positional argument 
# we can give any number of argument , even zero
def sum(*n):
    sum=0
    for i in n: #the input arguments are treated as tuple thats why i in n 
        sum+=i
    print(sum)

sum(1,2,3,4,5)
sum()
sum(-1,10,20,30)
```

```python
# variable length keyword argument 
# the arguments are treated as dictionary ie key value pair
def f1(**n):
    for k,v in n.items():
        print(k,v)
f1(name='bupender jogi', adress= 'bupender jogi', number='bupender jogi' )
```

```python
# lamda function
# single line function
# anonymous function 
# only one output
# syntax -- lamda variable : expression

# lambda function to return square of a number
p= lambda x : x**2
p(8)
```

```python
# lambda func to return sum of arguments
sum= lambda a,b,c,d : a+b+c+d
sum(1,2,3,4)
```

```python
# lf to multiply 2 numbers 
mul= lambda a,b: a*b
mul(2,5)
```

```python
# lf to add 2 numbers 
add= lambda a,b: a+b
add(2,5)
```

```python
# lambda func to return the greater number 
greater = lambda x,y : x if x>y else y # use of if statement in lambda 
greater(2,1)
greater(1,2)
```

```python
# lambda function to reveerse a string 
rev= lambda a: a[::-1]
rev('karan')
```

```python
# lambda to get the length of string 
length= lambda x : len(x)
length('karan')
```

```python
# can we give function as argument to other function ?
# what if we want to make specific change in a sequnce like square all the element of list
# we use map() function for this 
# maps chnages to each element in a collection 
def f(n):
    return n**2
s=list(map(f,[1,2,3,4]))
print(s)  
```

```python
# we can shorten the above code using lambda function as well
p= list(map(lambda x : x**2, [1,2,3,4]))
print(p)
```

```python
# we can use map and lambda function to multiply each element of two sequences
l1=[1,2,3,4]
l2=[4,3,2,1]
# we want to multiply the corresponding elements and return multiplied list 
ml=list(map(lambda x,y : x*y, l1,l2))
print(ml)
```

```python
# we can use map and lambda function to add each element of two sequences
l1=[1,2,3,4]
l2=[4,3,2,1]
ml=list(map(lambda x,y : x+y, l1,l2))
print(ml)
```

```python
# convert a list of strings to uppercase 
string=['hello','karan','sir']
Ustring= list(map(lambda x: x.upper(),string))
print(Ustring)
```

```python
# convert a list of numbers to string 
num=[1,2,3,4]
string= list(map(lambda x: str(x),num))
print(string)
```

```python
# convert a list of strings to list of lengths
string=['hello','karan','sir']
leng= list(map(lambda x: len(x),string))
print(leng)
```

```python
# convert a list of strings to list of initials
string=['hello','karan','sir']
initial= list(map(lambda x: x[0],string))
print(initial)
```

```python
# convert a list of strings to reverse list
string=['hello','karan','sir']
rev= list(map(lambda x: x[::-1],string))
print(rev)
```
