```python
#Static variable 
# can be accesed by any object 
# does not chnages obj to obj
# does not require intitialize and declare 

class test:
    a=10                             # static var delcared in class 
    def __init__(self):
        self.b=30                     #normal instance var delcared using self 
    def  m1(self):
        print(self.b)

t=test()
print(t.a)                         #acessed using class object name 

s=test()
print(s.a)                           # two objects using static  
```

```python
# static variables can be delcared using following ways
class test:
    a=10                            # static variable outside methods constructor but in class
    def __init__ (self):
        test.b=20                   # static variable in const but using classname 
    def m1(self):
        test.c=30                   # static variable in method but using classname 
    @classmethod
    def m2(cls):
        test.d=40                   # static variable in class method but using classname 
        cls.e=50                   # static variable in class method but using cls 
    @staticmethod
    def m3():
        test.f=70                  # static variable in static method but using classname 

t=test()
test.g=80                         # static variable declared outisde class uisng classname 
t.m1()
t.m2() 
t.m3()                             # call all methods to show static varibales in dict
print(test.__dict__)
```

```python
class test:
    a=10
    b=20
    c=30
    d=40
    e=50
    f=60
    g=80
    def __init__(self):
        print(test.a)
    def m1(self):
        print(test.b)
        print(test.c)
    @classmethod
    def m2(cls):
        print(test.d)
        print(cls.e)
    @staticmethod
    def m3():
        print(test.f)

t=test()
t.m1()
t.m2()
t.m3()
print(t.g)
```

```python
# Types of methods

#instance method 
# - uses instance variable
# - first paramter self neccesary 
# - var can be acess using self only 

class student:
    def __init__(self, name, rollno, marks):
        self.name=name
        self.rollno= rollno
        self.marks= marks
    def display(self):                                  #instance method as it uses self as first argument
        print(self.name)                                #instance method acceses the var delcared using constructor
        print(self.rollno)
    def grade(self):                                     # displaay() and grade() both are instance method 
        if self.marks>60:
            print('first class')
        elif self.marks>50:
            print('second class')
        elif self.marks>40:
            print('third class')
        else:
            print('failed')

#for multiple students usinf for loop

n=int(input('enter number of students'))
for i in range(n):
    name= input('enter name')
    rollno= int(input('enter roll no'))
    marks= int(input('enter marks'))
    s=student(name,rollno,marks)
    s.display()
    s.grade()
```

```python
# class method 
# the method which uses class variable ie static varibale is called class method
# needs decorator @classmethod
# needs cls as argument 

class test:
    legs=4 
    @classmethod                    #decorator
    def walk(cls,name):             # cls as arg, name is local var
        print(name,"walks with", cls.legs) # here leg is class variable or static varibale 

t= test()
t.walk('dog')
```

```python
# Static method 
# -- general utility method 
# -- uses only local var 
# -- not uses any static var or instance var of methods 
# -- not requires cls or self 
# -- can be accesed using obj reference or class name 

class test:
    @staticmethod
    def m1(name, age):
        print(f"hello {name} your age is {age}")

t=test()
t.m1('k',21)           #accesed using class obj reference
test.m1('k',21)        # accesed using class name only
```

```python
# Inheritance 
# child class inherits the properties of parent class 

class parent:                             
    def __init__(self, name , age):
        self.name= name
        self.age= age
    def display(self):
        print(self.name)
        print(self.age)

class child(parent):                  # child class created that inherits parent class
    pass

c= child('aj',22)                     # since its a child of parent it uses constructor of parent
c.display()                           # it can acess the methods of parent 
```

```python
# single inheritance 
# one child inherits from one parent 

class p:                                        # one parent class 
    def m1(self):
        print('this is parent method')

class c(p):                                      #one child class inherits one parent class
    def m2(self):
        print('this is child method')

C=c()
C.m1()           # acces own method 
C.m2()           # access parent mehtod 
```

```python
# multi level inheritance 
# multiple levels of inheritance 
# eg. c inherits b inherits a
# A --> B --> C

class a:
    def m1(self):                              #parent class a
        print('this is method of a ')

class b(a):
    def m2(self):
        print('this is method of b ')        # child b inherits a

class c(b):
    def m3(self):
        print('this is method of c ')       # child c inherits b, ie grandchild of a in slang

C=c()                            
C.m3()                                    # hence c can access methods of a ,b,c 
C.m2()
C.m1()
```
