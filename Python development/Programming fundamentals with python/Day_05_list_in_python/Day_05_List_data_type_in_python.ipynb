```python
# concept of Alias
# y is alias of z
z=[1,2,3,4,5,6]
y=z
print(id(z))
print(id(y))
z is y
```

```python
# concept of cloning
# using copy function
a=[1,2,3,4,5]
b=a.copy() # cloned the list but stored in different memory address
print(b)
print(id(a))
print(id(b))
```

```python
# cloning using slicing method
# still different memory location
a=[10,100,5,7,8,10]
b=a[::]
print(b)
print(id(a))
print(id(b))
```
