```python
# single level inheritance 
# class A inherits from B
class A:
    def m1(self):
        print('this is class A')

class B(A):
    def m2(self):
        print('this is class B child of A')

a= A()
b=B()
b.m2()
b.m1()
```

```python
# multi level inheritance
class A:
    def m1(self):
        print('this is class A')

class B(A):
    def m2(self):
        print('this is class B child of A')

class C(B):
    def m3(self):
        print('this is class C child of B')

c=C()
c.m3()
c.m2()
c.m1()
```

```python
# heirarchial inheritance 
# one parent class has multiple childs
#here  both B and C are child of A
class A:
    def m1(self):
        print('this is class A')

class B(A):
    def m2(self):
        print('this is class B child of A')

class C(A):
    def m3(self):
        print('this is class C child of A')

c=C()
c.m3()
c.m1()

b=B()
b.m2()
b.m1()
```

```python
# multiple inheritance
# A child class inherits from multiple parent class 
# here class C inherits from both A and B
class A:
    def m1(self):
        print('this is class A')

class B:
    def m2(self):
        print('this is class B')

class C(A,B):
    def m3(self):
        print('this is class C child of A and B')

c=C()
c.m3()
```

```python
# solving for a diagram
class A:
    def m1(self):
        print('this is class A')

class B(A):
    def m2(self):
        print('this is class B child of A')

class C(B):
    def m3(self):
        print('this is class C child of B')

class D:
    def m4(self):
        print('this is class D')

class E(B,D):
    def m5(self):
        print('this is class E child of D and B')

e=E()
e.m5()
e.m4()
e.m2()
e.m1()
```

```python
# # types of polymorphism 
# A) Overloading :
# 1.operator overloading (possible in python)
# 2.method overloading (not possible)
# 3.constructor overloading (not possible)
#  reason behind not possible is that python is dynamically types language

# B) Overriding :
# already declared properties of parent can be overidden by the child
# 1. method overiding
# 2. constructor overiding
```

```python
# example of operator overloading 
# adding contents of two classes 
class test:
    def __init__(self,pages):
        self.pages= pages
    def __add__(self, other):            # this special function is overloading + with its method
        return self.pages + other.pages

t1= test(100)
t2= test(200)

print(t1+t2) # normally addition of 2 objects not possible, but special function add overloads + with its function
        
```

```python
# example of operator overloading 
# multiplying contents of two classes

class employee:
    def __init__(self, name, pdsal):
        self.name= name
        self.pdsal= pdsal
    def __mul__(self,other):         # this special function is overloading * with its method
        return self.pdsal * other.days

class time_sheet:
    def __init__(self, name, days):
        self.name= name
        self.days= days

e= employee('k',500)
t= time_sheet('k',20) 

print(e*t) # normally multiplication of 2 objects not possible, but special function mul overloads + with its function
```

```python
# trying out if method overloading possible

class test:
    def m1(self):
        print('method with no argument') #python will take this as m1
    def m1(self, a):
        print('method with one arg') # py will take this as m1 and ignore the earlier one m1
    def m1(self,a,b):
        print('method with 2 arg') #pthon will define m1 with lastest declared method and ignore the earlier m1 and m2

t= test()
t.m1(1,2)
```

```python
# trying out if constructor overloading possible

class test:
    def __init__(self):
        print('method with no argument') #python will take this as const
    def __init__(self, a):
        print('method with one arg') # py will take this as const and ignore the earlier one
    def __init__(self,a,b):
        print('method with 2 arg') #pthon will define constructor with lastest declared and ignore the earlier

t= test(1,2)
```

```python
# example of overidding (method overriding)
class parent:
    def marriage(self):
        print('marry R')   #parent defined method marriage

class child(parent):
    def marriage(self):
        print('marry A')  # child overidens marriage method

c=child()
c.marriage()
```

```python
# if in above we want to run both parent and child method 
# use of super keyword 
class parent:
    def marriage(self):
        print('marry R')   #parent defined method marriage

class child(parent):
    def marriage(self):
        super().marriage()
        print('marry A')  # child overidens marriage method but still calls parent method using super keyword

c=child()
c.marriage()
```

```python
# constructor overiding 
class parent:
    def __init__(self,name,age):
        self.name= name
        self.age= age
        print('parent constructor ran')

class child(parent):
    def __init__(self,sal):
        self.sal=sal
        print('child constructor ran')  # the child will overide the parent constructor and use its own constructor  

c= child(500)
```

```python
# if child wants to use both parent and child constructor
# use of super keyword
class parent:
    def __init__(self,name,age):
        self.name= name
        self.age= age
        print('parent constructor ran')

class child(parent):
    def __init__(self,sal,name,age):
        super().__init__(name,age)
        self.sal=sal
        print('child constructor ran')  # the child will overide the parent constructor but call parent constructor as well bcz of super keyword 

c= child(500,'k',21)
```

```python
# abstraction 
# to show only the neccesary things and hide unnesccesary implementation 
# ABC class from abc library is required
# also decorator is used to declare a abstract method
from abc import * # we will use (abstract method, ABC class)
class test(ABC):
    @abstractmethod
    def m1(self):
        print('hi')

t=test() # will return error since test is abstract class cannot have object
```

```python
from abc import * # we will use (abstract method, ABC class)
class test(ABC):
    def m1(self): # didnt mentioned abstract method
        print('hi')

t=test()  #object created 
```

```python
from abc import * # we will use (abstract method, ABC class)
class test():     # didnt inherited from ABC class
    @abstractmethod
    def m1(self):
        print('hi')

t=test() #object created 
```
