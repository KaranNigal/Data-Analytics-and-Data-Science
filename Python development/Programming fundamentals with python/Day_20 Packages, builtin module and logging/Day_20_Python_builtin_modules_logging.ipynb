```python
# loging module in python
# - used to maintain logs of a exucation of program 
# - helps  in debugging
# level name -- value
# debug - 10
# info - 20
# warning - 30
# error - 40
# critical - 50
# basicConfig is used to configure the log file
# if we select a level of logging all the levels below it will be checked ie, if level is info then from info to critical
```

```python
import logging

logging.basicConfig(filename='applog.log',level=logging.INFO,filemode='w') #this applog file will be created with write mode and level info and higher
print('Logging demo')
logging.debug('debug info')
logging.info('info info')
logging.warning('warning info')
logging.error('error info')
logging.critical('critical info')
```

```python
# to show logs on terminal only remove filename and mode
# restart the kernel to run log file again
import logging

logging.basicConfig(level=logging.INFO)
print('Logging demo')
logging.debug('debug info')
logging.info('info info')
logging.warning('warning info')
logging.error('error info')
logging.critical('critical info')
```

```python
#adding the format paramter to custom the output of logs
import logging

logging.basicConfig(level=logging.INFO, format='%(asctime)s-%(levelname)s-%(message)s') #this applog file will be created with write mode and level info and higher
print('Logging demo')
logging.debug('debug info')
logging.info('info info')
logging.warning('warning info')
logging.error('error info')
logging.critical('critical info')
```
