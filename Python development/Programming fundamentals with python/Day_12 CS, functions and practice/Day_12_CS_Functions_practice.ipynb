```python
# wf to return even numbers from a range 
def even(a,b):
    l=[i for i in range(a,b+1) if i%2==0]
    print(l)
even(1,20)
```

```python
# wf to return odd numbers from a range 
def odd(n):
    for i in range(n+1):
        if i%2!=0:
            print(i)
odd(10)
```

```python
# function to display the factorial of given number
n=int(input('Enter the number to display the factorial'))
def factorial(n):
    f=1
    for i in range(1,n+1):
       f= f*i
    return f
print(factorial(n))
           
```

```python
# function to check palindrome
def is_palindrome(s):
    if s==s[::-1]:
        return True
    else:
        return False
is_palindrome('madam')
```
