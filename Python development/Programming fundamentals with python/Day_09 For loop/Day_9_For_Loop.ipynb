```python
# to check if the given string is palindrome using for loop
s= input("Enter the string to check if palindrome")
result= True
for i in range(len(s)):
    if(s[i]!=s[-(i+1)]):
        result= False
        break
print(result)
```

```python
# alternate sol for # to check if the given string is palindrome using for loop
s=input("Enter the string")
p=''
for i in s:
    p=i+p
print(p)
if s==p:
    print('palindrome')
else:
    print('not palindrome')
```

```python
# print factorial of a given number 
n=int(input("Enter the number to find the factorial of"))
fact= 1
for i in range(1,n+1):
    fact*= i
print(fact)
```

```python
# else block in for loop
for i in range(10):
    print(i)
else:
    print("all loops ran successfuly")
```

```python
# else block in for loop with break
for i in range(10):
    if i==6:
        break
    print(i)
else:
    print("all loops ran successfuly")
# Hence else block runs only when complete for loop is run successfuly otherwise does not executed
# used to verify wheater complete loop is been run or not 
```

```python
# else block in for loop with continue
for i in range(10):
    if i==6:
        continue
    print(i)
else:
    print("all loops ran successfuly")
# if continue keyword then else runs because loops gets completed  
```
